```python
from elasticsearch import Elasticsearch

es = Elasticsearch("http://localhost:9200", basic_auth=("elastic", "elastic"))

# Create the index
index_name = "fifa_laws"
if not es.indices.exists(index=index_name):
    es.indices.create(index=index_name, body={
        "mappings": {
            "properties": {
                "title": {"type": "text"},
                "content": {"type": "text"}
            }
        }
    })

# Add documents to the index
documents = [
    {
        "title": "Functions and Composition of the Football Tribunal",
        "content": "The Football Tribunal is composed of..."
    },
    # Add more documents as needed
]

for doc in documents:
    es.index(index=index_name, document=doc)

print(f"Index '{index_name}' created and populated with documents.")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Define model name and local save path
model_name = "Qwen/Qwen2.5-7B-Instruct"
save_path = r"C:\Users\mouni\OneDrive\Bureau\rag_app_deployment\rag-app\model1"  # Change this to your desired local path

# Load model and tokenizer
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,  # Adjust dtype as needed
    device_map="auto"
)

tokenizer = AutoTokenizer.from_pretrained(model_name)

# Save to local path
model.save_pretrained(save_path)
tokenizer.save_pretrained(save_path)

print(f"Model and tokenizer saved to {save_path}")
```
